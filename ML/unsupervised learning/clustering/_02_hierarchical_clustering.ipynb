```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import datasets
```

```python
iris = datasets.load_iris()
```

```python
iris_data = pd.DataFrame(iris.data)
```

```python
iris_data.columns=iris.feature_names
```

```python
iris_data
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()  # Create the scaler object
X_scaled = scaler.fit_transform(iris_data)  # Fit and transform the data
```

```python
X_scaled
```

```python
X_scaled.shape
```

```python
from sklearn.decomposition import PCA
```

```python
pca =PCA(n_components=2)
```

```python
pca
```

```python
pca_scaled=pca.fit_transform(X_scaled)
```

```python
plt.scatter(pca_scaled[:,0],pca_scaled[:,1])
```

```python
pca_scaled.shape
pca_scaled
```

```python
#to construct dendogram
import scipy.cluster.hierarchy as sc
```

```python
#plot the dendogram 
plt.figure(figsize=(50,40))
plt.title("Dendograms")
```

```python
#create dendogram
sc.dendrogram(sc.linkage(pca_scaled, method='ward'))
plt.title('Dendogram')
plt.xlabel('sample index')
plt.ylabel('Eucledian Distance')
```

```python
from sklearn.cluster import AgglomerativeClustering 
```

```python
cluster = AgglomerativeClustering(n_clusters=2, metric='euclidean', linkage='ward')
cluster.fit(pca_scaled)
```

```python
# Access the cluster labels
cluster.labels_
```

```python
plt.scatter(pca_scaled[:,0],pca_scaled[:,1],c=cluster.labels_)
```

```python
from sklearn.metrics import silhouette_score

# Fit the model
cluster = AgglomerativeClustering(n_clusters=2, metric='euclidean', linkage='ward')
cluster_labels = cluster.fit_predict(pca_scaled)

# Calculate silhouette score
score = silhouette_score(pca_scaled, cluster_labels)
print(f"Silhouette Score: {score}")
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score

# Range of cluster numbers
range_n_clusters = range(2, 11)
silhouette_scores = []

for n_clusters in range_n_clusters:
    cluster = AgglomerativeClustering(n_clusters=n_clusters, metric='euclidean', linkage='ward')
    labels = cluster.fit_predict(pca_scaled)
    score = silhouette_score(pca_scaled, labels)
    silhouette_scores.append(score)

# Plotting the Elbow Graph
plt.plot(range_n_clusters, silhouette_scores, marker='o')
plt.title('Elbow Graph (Silhouette Score)')
plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Score')
plt.grid(True)
plt.show()
```
