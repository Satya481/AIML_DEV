```python
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
import numpy as np
```

```python
# Step 1: Input data (age) and labels (bought or not)
X = [[22], [25], [47], [52], [46], [56], [23]]
y = [0, 0, 1, 1, 1, 1, 0]
```

```python
# Step 2: Train the model
model = LogisticRegression()
model.fit(X, y)
```

```python
age = [[30]]
prediction = model.predict(age)
print("Prediction for age 30:", prediction[0])
```

```python
# Step 4: Visualization
ages = np.arange(20, 60, 1).reshape(-1, 1)  # ages 20 to 59
predicted_probs = model.predict_proba(ages)[:, 1]  # get probability of "buying"
```

```python
plt.figure(figsize=(10, 6))
plt.scatter([age[0] for age in X], y, color='blue', label='Training Data')
plt.plot(ages, predicted_probs, color='red', label='Prediction Curve')
plt.axvline(x=30, color='green', linestyle='--', label='Test Age: 30')
plt.title("Will a person buy based on age?")
plt.xlabel("Age")
plt.ylabel("Probability of Buying")
plt.legend()
plt.grid(True)
plt.show()
```
