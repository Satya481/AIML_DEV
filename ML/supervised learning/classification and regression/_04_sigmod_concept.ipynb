```python
import numpy as np
import matplotlib.pyplot as plt

# Sigmoid function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# X values from -10 to 10
z = np.linspace(-10, 10, 100)
probabilities = sigmoid(z)

# Plotting
plt.figure(figsize=(8, 4))
plt.plot(z, probabilities, color='green', linewidth=2)

# Highlight the 0.5 probability line
plt.axhline(y=0.5, color='gray', linestyle='--', label='50% Probability')

# Highlight the decision point (z = 0)
plt.axvline(x=0, color='blue', linestyle='--', label='z = 0 (Decision Boundary)')

# Labels and titles
plt.title('Sigmoid Curve in Logistic Regression')
plt.xlabel('z = w1*x + b')
plt.ylabel('Probability')
plt.legend()
plt.grid(True)
plt.show()
```
