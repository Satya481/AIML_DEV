# Arithmetic Function

```python
import numpy as np
arr = np.array([1,2,3,8,5,4])
print("min: ",np.min(arr), np.argmin(arr))   # to get the minimum value from the array. argmin--> to get the index at where it is present
print("max: ",np.max(arr), np.argmax(arr))   #to get the maximum value from the array. argmin--> to get the index at where it is present
```

```python
var_1 = np.array([[2,3,5],[7,9,3]])
print(np.min(var_1, axis=1))  # axis=1 for row
print(np.min(var_1, axis=0)) # axis=0 for column
```

```python
# sqrt() for squareroot 
var_2 = np.array([1,2,3,8,5,4])
print(np.sqrt(var_2))
```

```python
# sin() for sine value of matrix
var_2 = np.array([1,2,3,8,5,4])
print(np.sin(var_2))
```

```python
# cos() for sine value of matrix
var_2 = np.array([1,2,3,8,5,4])
print(np.cos(var_2))
```

```python
# cumsum() for sine value of matrix
var_2 = np.array([1,2,3,8,5,4])
print(np.cumsum(var_2)) # cumsum is a function like 1+2=3, 3+3=6, 6+8=14,.....
```
