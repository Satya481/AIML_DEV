# search

```python
import numpy as np
var = np.array([1,2,3,4,5,6,4,5,2,3,2])
print(np.where(var==2))
               
```

```python
var1 = np.array([1,2,3,4,6,7])
print(np.searchsorted(var1,5))
```

# sort

```python
var3 = np.array([4,64,62,654,497,662,484,23,25484,66,4612,6255,92323,64,412121])
print(np.sort(var3))
```

# shuffle

```python
var4 = np.array([1,2,3,4,5])
np.random.shuffle(var4)
print(var4)
```

# unique

```python
var5 = np.array([1,2,3,2,5,6,5])
x11 = np.unique(var5,return_index=True)
print(x11)
```

# resize

```python
var6 = np.array([1,2,3,4,5,6])
print(np.resize(var6,(3,2)))
```
