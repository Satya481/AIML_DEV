# Indexing in numpy 

```python
import numpy as np
var_1 = np.array([1,2,3])
print(var_1[1])
```

# 2D array

```python
var_2 = np.array([[1,2],[4,5]])                       #     0  1
print(var_2[1,1])                                     #    [[1 2] --> 0  
                                                      #     [4 5]]--> 1
```
