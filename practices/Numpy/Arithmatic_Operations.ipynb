# Arithmatic Operations

 # 1. Add Operation

```python
import numpy as np 
arr_1 = np.array([[1,2,3,4],[5,6,4,9]])
arr_1_new = arr_1 + 3
print(arr_1_new)
print(arr_1.ndim)
```

```python
arr_1 = np.array([[1,2,3,4],[5,6,4,9]])
arr_2 = np.array([[5,7,6,4],[1,5,45,6]])
arr_new = arr_1 + arr_2
print(arr_new)
print(arr_new.ndim)
```

```python
arr_1 = np.array([[1,2,3,4],[5,6,4,9]])
mul= arr_1 * 3
print(mul)
```

```python
arr_1 = np.array([[1,2,3,4],[5,6,4,9]])
arr_2 = np.array([[5,7,6,4],[1,5,45,6]])
arr_new = arr_1 * arr_2
print(arr_new)
```

```python
# Define two 2D matrices
matrix_a = np.array([[1, 2], [3, 4]])
matrix_b = np.array([[5, 6], [7, 8]])

# Perform matrix multiplication
result = np.matmul(matrix_a, matrix_b)

print(result)
```
