# 1.  ARRAY

```python
import numpy as np 
arr=[1,2,3,4]
y=np.array(arr)
print(y)
print(type(y))
print(y.ndim)
```

# 3.  Conversion to multi dimensional array

ndim 
ndmin

```python
import numpy as np
array_5=np.array([1,2,3,5,8],ndmin=10) # ALERT IT IS 'ndmin' not 'ndim'
print(array_5)
print(array_5.ndim)
```
