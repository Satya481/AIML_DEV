# RANDOM

 # 1. rand() ( to create random number from 0 to 1)

```python
import numpy as np
var = np.random.rand(4)
print(var)
```

```python
var_2 = np.random.rand(2,3)
print(var_2)
```

# 2. randn() (to create random number close to zero it can be positive or nagetive)

```python
var_3 = np.random.randn(4)
print(var_3)
```

# 3. ranf() (to generate random float numbers from 0.0 --> 1.0(tends to 1))

```python
var4 = np.random.ranf(4)
print(var4)
```

# 4. randint() (to generate random number to given range)

```python
var5 = np.random.randint(1,9,(4,5)) # (min=1, max=9,(row=4,column=5))
print(var5)
```
