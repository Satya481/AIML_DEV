```python
import numpy as  np
var1 = np.array([1])
print("Data type: ", var1.dtype)
```

```python
var2 = np.array([1.0,2.1,3.5])
print(type(var2), var2.dtype)
```

```python
var2 = np.array(["s","S","y"])
print(type(var2), var2.dtype) # string data type is 'U1'
```

```python
var2 = np.array(["s","S","y",'d',1,5,6,5])
print(type(var2), var2.dtype) # string data type is 'U21'
```

```python
x = np.array([1,2,3,25], dtype = np.int8)
print(x.dtype)  
```

```python
x = np.array([1,2,3,25], dtype = "f") # convertion of data-type using short-cut characters
print(x.dtype)
print(x)
```

```python
x = np.array([1,2,3,25]) 
print(x.dtype)
print(x)
new = np.float64(x)
print(new.dtype)
print(new)
```
